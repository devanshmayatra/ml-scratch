```python
import numpy as np
```

```python
def mahalanobis_sq(x, mean, variances):
    Xdist = x[0] - mean[0]
    Ydist = x[1] - mean[1]
    Xsd = np.sqrt(variances[0])
    Ysd = np.sqrt(variances[1])
    
    X_sd_deviation = Xdist/Xsd
    Y_sd_deviation = Ydist/Ysd
    
    d2 = (X_sd_deviation**2) + (Y_sd_deviation**2)
    
    return d2
```

```python
mean = np.array([-3, 2])
variances = np.array([4, 16])

print(mahalanobis_sq(np.array([-1, 6]), mean, variances))  # 2
print(mahalanobis_sq(np.array([-3, 10]), mean, variances)) # 4
```
