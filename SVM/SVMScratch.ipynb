```python
class SVMScratch:
  def __init__(self, learning_rate=0.001, lambda_param=0.01, n_iters=1000):
    self.lr = learning_rate
    self.lambda_param = lambda_param
    self.n_iters = n_iters
    self.w = None
    self.b = None
    
  def fit(self,X,y):
    n_samples = len(X)
    n_features = len(X[0])
    
    y_ = [-1 if i <= 0 else 1 for i in y]
    
    self.w = [0.0] * n_features
    self.b = 0.0
    
    #Training loop
    for _ in range(self.n_iters):
      for i in range(n_samples):
        
        
        #calculate decision values x.w - b
        decision_value = 0.0
        for j in range(n_features):
          decision_value +=self.w[j] * X[i][j]
        decision_value -= self.b
        
        #check hinge condition
        condition = y_[i] * decision_value >=1
        
        if condition:
          
          #Case A: no misclassification
          #Only apply regularization
          # w = w - lr * (2 * lambda_param * w)
          
          for j in range (n_features):
            self.w[j] -=self.lr * (2 * self.lambda_param * self.w[j])
            
        else:
        #Case B: misclassification
        # w = w - lr * (2 * lambda_param * w - y_i * x_i)
        # b = b - lr * (-y_i)
        
          for j in range(n_features):
            grad = (2 * self.lambda_param * self.w[j]) - (y_[i] * X[i][j])
            self.w[j] -= self.lr * grad
          self.b -= self.lr * y_[i]
            
  def predict(self,X):
    predictions = []
    for i in range(len(X)):
      decision_val = 0.0
      for j in range (len(X[0])):
        decision_val +=self.w[j] * X[i][j]
      decision_val -= self.b
      
      predictions.append(1 if decision_val >=0 else -1)
    return predictions
```

```python
# --- Usage Example ---
if __name__ == "__main__":
    # Simple data:
    # Points [1,1], [2,2] are Class 1
    # Points [5,5], [6,6] are Class -1
    X = [[1, 1], [2, 2], [5, 5], [6, 6]]
    y = [1, 1, -1, -1]

    clf = SVMScratch()
    clf.fit(X, y)

    # Test [1.5, 1.5] -> Should be 1
    # Test [7, 7] -> Should be -1
    test_X = [[1.5, 1.5], [7, 7]]
    print("Predictions:", clf.predict(test_X))
```
