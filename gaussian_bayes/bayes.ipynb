```python
import numpy as np
```

```python
def calc_posterior(priors=[], likelihoods=[]):
    scores = priors*likelihoods
    posteriors = []
    pe = scores.sum()
    
    if pe == 0:
        raise ValueError("Posterior is undefined because evidence probability is zero.")
    
    for ele in scores:
        posteriors.append(ele/pe)
        
    return f"Scores: {scores}, Evidence Probability: {pe}, Posteriors: {posteriors}"
```

```python
priors = np.array([0.3, 0.7])
likelihoods = np.array([0.1, 0.05])

calc_posterior(priors,likelihoods)
```

```python
priors = np.array([0.2, 0.5, 0.3])
likelihoods = np.array([0.6, 0.2, 0.4])

calc_posterior(priors,likelihoods)
```

```python
priors = np.array([0.1, 0.3, 0.6])
likelihoods = np.array([0.5, 0.5, 0.5])

calc_posterior(priors,likelihoods)
```
