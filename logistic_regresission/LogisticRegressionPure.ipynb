```python
import math

class LogisticRegressionPure:
  def __init__(self,learning_rate=0.01, iterations=1000):
    self.lr = learning_rate
    self.iterations = iterations
    self.weights = []
    self.bias = 0.0
    
  def _sigmoid(self,z):
    if z < -100: return 0
    if z > 100: return 1
    
    return 1 / ( 1 + math.exp(-z))
  
  def fit(self, X, y):
    n_samples = len(X)
    n_features = len(X[0])
    
    self.weights = [0.0] * n_features
    self.bias = 0.0
    
    for _ in range(self.iterations):
      
      linear_predictions = []
      predictions = []
      
      for i in range(n_samples):
        z = 0.0
        
        for j in range(n_features):
          z += X[i][j] * self.weights[j]
        z +=self.bias
        
        pred = self._sigmoid(z)
        predictions.append(pred)
        
        dw = [0.0] * n_features
        db = 0.0
        
      for i in range(n_samples):
        error = predictions[i] - y[i]
        
        db +=error
        for j in range(n_features):
          dw[j] += error * X[i][j]
          
      db /= n_samples
      for j in range(n_features):
        dw[j] = dw[j] / n_samples
      
      self.bias -= self.lr * db
      for i in range(n_features):
        self.weights[j] += self.lr * dw[i]
        
  def predict_proba(self,X):
    """
    Returns probability from 0 to 1
    """
    
    probs = []
    for i in range(len(X)):
      z = 0.0
      for j in range(len(X[0])):
        z +=X[i][j] * self.weights[j]
      z += self.bias
      probs.append(self._sigmoid(z))
    return probs
  
  def predict(self,X,threshold = 0.5):
    """Returns Class 0 or 1 based on threshold"""
    probs = self.predict_proba(X)
    return [1 if p >= threshold else 0 for p in probs]
```

```python
# --- Usage Example (XOR Logic -ish) ---
if __name__ == "__main__":
    # Data: 2 Features.
    # [0, 0] -> 0
    # [1, 1] -> 1
    # [2, 2] -> 1
    # [0, 2] -> 0
    X_train = [[0, 0,0], [1, 1,0], [2, 2,0], [0, 2,1]]
    y_train = [0, 1, 1, 0]

    model = LogisticRegressionPure(learning_rate=0.1,iterations=5000)
    model.fit(X_train, y_train)

    test = [[0.1, 0.1], [1.5, 1.5]]
    print("Probabilities:", model.predict_proba(test))
    print("Predictions:", model.predict(test))
```
